En el bloque 1.1 defino el propósito del chatbot, donde especifico que está orientado a la atención de estudiantes universitarios y tambien describiendo su funcionalidad principal.

En este bloque construyo el dataset del chatbot mediante pares de preguntas y respuestas.

##Separacion de Preguntas

```python
preguntas = [par[0] for par in qa_pairs]
respuestas = [par[1] for par in qa_pairs]

for i in range(len(qa_pairs)):
    print(f"Pregunta {i+1}: {preguntas[i]}")
    print(f"Respuesta {i+1}: {respuestas[i]}")
    print("-" * 50)
```

En este bloque lo que hago es mostrar las preguntas y respuestas del dataset en formato vertical, lo que permite una mejor lectura.

###  2) Crear el chatbot utilizando TFIDF y similitud del coseno. (1 punto)



```python
#Vectorizo las preguntas (TF-IDF)

#Importa la herramienta que convierte texto en vectores
from sklearn.feature_extraction.text import TfidfVectorizer
#Importa la función que después va a comparar preguntas
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()

# Convertimos las preguntas en vectores
tfidf_matrix = vectorizer.fit_transform(preguntas)
```

En este bloque utilizo TF-IDF para transformar las preguntas en vectores numéricos, lo que permite representar el texto de forma matemática y poder compararlo.

```python
#Función del chatbot

def chatbot_respuesta(pregunta_usuario):
    # Convertir la pregunta del usuario en vector
    pregunta_vector = vectorizer.transform([pregunta_usuario])

    # Calcular similitud del coseno
    similitudes = cosine_similarity(pregunta_vector, tfidf_matrix)

    # Obtener índice de la pregunta más similar
    indice = similitudes.argmax()

    # Devolver la respuesta correspondiente
    return respuestas[indice]
```

En este bloque defino la función del chatbot, que transforma la pregunta del usuario en un vector y calcula su similitud con las preguntas del dataset para encontrar la más parecida y devolver su respuesta.

```python
#Pruebo el chatbot

while True:
    user_input = input("Tú: ").lower().strip()

    # Saludos
    if any(palabra in user_input for palabra in ["hola", "buenas", "buen dia", "buenas tardes"]):
        print("Chatbot: ¡Hola!  ¿En qué puedo ayudarte?")
        continue

    # Salida
    if any(palabra in user_input for palabra in ["salir", "chau", "adios", "gracias", "no tengo mas dudas", "listo"]):
        print("Chatbot: ¡Perfecto! Si tienes otra duda, solo pregúntame ")
        break

    # Respuesta normal
    respuesta = chatbot_respuesta(user_input)
    print("Chatbot:", respuesta)
```

### 3) Crear otro chatbot utilizando embeddings. Indique cuál embedding (1 punto) pre-entrenado eligió.



```python
!pip install gensim --quiet
import gensim.downloader as api

# Cargar modelo pre-entrenado
modelo = api.load("glove-wiki-gigaword-50")
```

En este bloque se carga un modelo de embeddings pre-entrenado (GloVe), el cual permite representar palabras como vectores numéricos basados en su significado semántico.

```python
#Función para convertir texto a vector

import numpy as np

def texto_a_vector(texto):
    palabras = texto.lower().split()
    vectores = []

    for palabra in palabras:
        if palabra in modelo:
            vectores.append(modelo[palabra])

    if len(vectores) == 0:
        return np.zeros(modelo.vector_size)

    return np.mean(vectores, axis=0)
```

En este bloque se transforma cada texto en un vector promedio de sus palabras, utilizando el modelo de embeddings, lo que permite representar el significado general de la oración.

```python
#Vectorizar todas las preguntas

vectores_preguntas = [texto_a_vector(p) for p in preguntas]
```

Vectorizo todas las preguntas del dataset en vectores utilizando embeddings

```python
#Funcion del Chatbot

from sklearn.metrics.pairwise import cosine_similarity

def chatbot_embeddings(pregunta_usuario):
    vector_usuario = texto_a_vector(pregunta_usuario)

    similitudes = cosine_similarity([vector_usuario], vectores_preguntas)

    indice = similitudes.argmax()

    return respuestas[indice]
```

En este bloque se calcula la similitud del coseno entre la pregunta del usuario y las preguntas del dataset utilizando embeddings

```python
#prueba del Chatbot

while True:
    user_input = input("Tú: ").lower().strip()

    if any(palabra in user_input for palabra in ["salir", "chau", "adios", "gracias", "no tengo mas dudas"]):
        print("Chatbot: ¡Perfecto! Si tienes otra duda, solo pregúntame ")
        break

    if any(palabra in user_input for palabra in ["hola", "buenas"]):
        print("Chatbot: ¡Hola!  ¿En qué puedo ayudarte?")
        continue

    respuesta = chatbot_embeddings(user_input)
    print("Chatbot:", respuesta)
```
